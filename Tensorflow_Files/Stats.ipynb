```python
import statistics
import numpy as np
from scipy import stats

# Sample data
data = [12, 15, 18, 22, 25, 25, 30, 35, 40, 45]

# ---------------------------
# Using built-in statistics module
# ---------------------------
print("Mean:", statistics.mean(data))
print("Median:", statistics.median(data))
print("Mode:", statistics.mode(data))
print("Variance:", statistics.variance(data))       # sample variance
print("Standard Deviation:", statistics.stdev(data)) # sample std dev
```

```python
# ---------------------------
# Using NumPy
# ---------------------------
arr = np.array(data)

print("\nNumPy Mean:", np.mean(arr))
print("NumPy Median:", np.median(arr))
print("NumPy Std Dev:", np.std(arr))         # population std dev by default
print("NumPy Std Dev (sample):", np.std(arr, ddof=1))
print("NumPy Variance:", np.var(arr))
print("Min:", np.min(arr))
print("Max:", np.max(arr))
print("Range:", np.max(arr) - np.min(arr))
```

```python
# Percentiles / Quartiles
print("25th percentile (Q1):", np.percentile(arr, 25))
print("50th percentile (Median):", np.percentile(arr, 50))
print("75th percentile (Q3):", np.percentile(arr, 75))
```

```python
# ---------------------------
# Using SciPy (more advanced stats)
# ---------------------------
print("\nSkewness:", stats.skew(arr))
print("Kurtosis:", stats.kurtosis(arr))
print("Z-scores:", stats.zscore(arr))

# Descriptive summary
desc = stats.describe(arr)
print("\nSciPy Describe:")
print(desc)
```

```python
# ---------------------------
# Frequency / Counts
# ---------------------------
from collections import Counter
freq = Counter(data)
print("\nFrequency count:", freq)
```
